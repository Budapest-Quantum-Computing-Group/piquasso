## Állítás

Legyen (2N) logikai módus, (N) páros, a módusindexek

[
0,1,\ldots,2N-1.
]

Tegyük fel, hogy a rectangular háló páros réteggel kezdődik és páratlan réteggel végződik, ahol

[
L_{\mathrm e}:\quad
(0,1),(2,3),\ldots,(2N-2,2N-1),
]

[
L_{\mathrm o}:\quad
(1,2),(3,4),\ldots,(2N-3,2N-2).
]

A Bell-sMZI (\to) MBLO-(V) átírás minden MZI után egy (R(\pi)) korrekciót eredményez. Ezek a korrekciók úgy rendezhetők át, hogy

[
\boxed{\text{a bulk páros rétegek előtt nem marad külön }\pi\text{-rotáció,}}
]

míg minden páratlan réteg korrekciója ekvivalens azzal, hogy

* minden MZI-jén egy közös (G(\pi)) rotáció van, és
* az alsó szélső (2N-1)-edik móduson egy (R_{2N-1}(\pi)) van.

Ez utóbbi pedig

[
\boxed{
\left(
\prod_{(j,j+1)\in L_{\mathrm o}}G_{j,j+1}(\pi)
\right)
R_{2N-1}(\pi)
=============

\mathcal R(\pi)R_0(-\pi),
}
]

ahol

[
\mathcal R(\pi)
===============

\prod_{j=0}^{2N-1}R_j(\pi)
]

az összes módus közös (\pi)-rotációja.

Ha a hálóban páros számú páratlan réteg van, akkor az összes (\mathcal R(\pi)) faktor egzaktul kioltódik, ezért a teljes átalakítás **nemcsak globális fázisig, hanem pontosan** ekvivalens azzal, hogy a megfelelő páratlan rétegeknél csak a felső szélső móduson hajtunk végre (R_0(-\pi))-t.

### 1. Páratlan réteg

A páratlan réteg MZI-jei

[
(1,2),(3,4),\ldots,(2N-3,2N-2),
]

így az sMZI(\to V) átalakításból származó korrekciók

[
R_2R_4R_6\cdots R_{2N-2}.
]

Párosítsuk őket

[
R_{4n+2}R_{4n+4}
================

G_{4n+2}G_{4n+3}.
]

Itt

[
G_{4n+3}=G_{4n+3,4n+4}
]

maga a páratlan réteg egyik MZI-jére esik, míg

[
G_{4n+2}=G_{4n+2,4n+3}

a szomszédos páros réteg MZI-jére esik.

[

[
R_{2N-2}
========

G_{2N-2,2N-1}R_{2N-1}.
]

A (G_{2N-2,2N-1}) szintén páros-layer típusú, míg

[
R_{2N-1}
]

az alsó edge-en közvetlenül végrehajtható.

[
\boxed{

[
\boxed{
G_{4n+2,,4n+3}

[

## Több páratlan réteg

Ha összesen (K) páratlan réteg van, akkor

[

===================

\mathcal R(\pi)R_0^{(r)}(-\pi),
]

ahol az (R_0^{(r)}) természetesen az (r)-edik páratlan réteg megfelelő longitudinális helyén található.

Mivel (\mathcal R(\pi)) közös phase rotation, minden passzív linear-optical gate-tel kommutál. Ezért az összes ilyen faktor összegyűjthető:

[
\mathcal R(\pi)^K.
]

A Fock-téren

[
\mathcal R(\pi)
===============

e^{i\pi\hat N_{\mathrm{tot}}},
]

így

[
\mathcal R(\pi)^2
=================

# e^{i2\pi\hat N_{\mathrm{tot}}}

I.
]

Ennélfogva

[
\boxed{
K\ \text{páros}
\quad\Longrightarrow\quad
\mathcal R(\pi)^K=I.
}
]

Tehát páros számú páratlan layer esetén az ekvivalencia **egzakt az egész Fock-téren**:

[
\boxed{
\text{eredeti }\pi\text{-korrekciók}
====================================

\prod_{r=1}^{K}R_0^{(r)}(-\pi),

}

[
\mathcal R(\pi)

marad vissza; interferométermátrix-szinten ez (-I), de általános CV/Fock-állapoton ez total-number parity, ezért ott nem szabad puszta globális fázisként eldobni.

A (2N)-módusú, (N) páros, **even (\to) odd** rectangular hálóban a (\pi)-korrekciók checkerboard szerkezete miatt:

[

[
\boxed{

[

We choose the propagation direction such that all such residual phases terminate on the same boundary mode; this is also the convention adopted by Strawberry Fields' `rectangular_compact` implementation.
